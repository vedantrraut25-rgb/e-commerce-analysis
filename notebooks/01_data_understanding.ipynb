# handling duplicated value

```python
print(f"CUSTOMER ID CHECK:\nRows: {len(customers)}\nUnique ID {customers['customer_id'].nunique()}\nDuplicate ID {customers['customer_id'].duplicated().sum()}  ")

print(f"\nORDER ID CHECK\nRows: {len(orders)}\nUnique ID {orders['order_id'].nunique()}\nDuplicate ID {orders['order_id'].duplicated().sum()} ")


print(f"\nPRODUCT ID CHECK\nRows: {len(products)}\nUnique ID {products['product_id'].nunique()}\nDuplicate ID {products['product_id'].duplicated().sum()}")


print(f"\nORDER ITEM CHECK\nRows: {len(order_items)}\nUnique ID {order_items['order_id'].nunique()}\nDuplicate ID {order_items['order_id'].duplicated().sum()}")
print("Rows:", len(order_items))
print(
    "Duplicate order-product pairs:",
    order_items.duplicated(
        subset=["order_id", "product_id"]
    ).sum()
)
```
